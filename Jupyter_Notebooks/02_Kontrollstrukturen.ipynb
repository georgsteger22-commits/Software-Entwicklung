## 1. Verzweigungen und Schleifen

### 1.1. Sequenz

Sequenzen bildet man in Python, indem man einfach Befehle untereinander schreibt. Damit Python einzelne Befehlsblöcke auseinanderhalten kann, muss man darauf achten, dass die Befehle innerhalb eines Blocks gleich eingerückt sind (Genaueres siehe unten).

### 1.2. Verzweigung

Die Verzweigung schaut in Python folgendermaßen aus:

```Python
if <Bedingung B1>:
    <Anweisungsblock A1>
elif <Bedingung B2>:
    <Anweisungsblock A2>
else:
    <Anweisungsblock A3>
```

Der `elif` Zweig ist optional, und kann auch mehrfach vorkommen. Auch der `else`-Zweig ist optional.

Wichtig zu beachten bei den Anweisungsblöcken ist, dass diese im Vergleich zum umliegenden Code eingerückt sind. Python erkennt anhand der Einrückung, welche Befehle Teil der Verzweigung sind, und welche schon zum nachfolgenden Code gehören. Welche Zeichen für die Einrückung verwendet werden (Leerzeichen oder Tab) und wie viele Zeichen eingerückt wird, ist frei wählbar. Wichtig ist nur, dass die Einrückungen innerhalb eines Blocks einheitlich sind. Daher nicht Leerzeichen und Tab mischen, oder manche Befehle unterschiedlich viele Zeichen einrücken. Darauf wird Python mit einer Fehlermeldung reagieren.

````python
# Wir definieren uns eine Variable, die wir in den folgenden Beispielen verwenden können
a = 3
````

````python
if a < 4:                               # Hier bitte den Doppelpunkt am Ende der if-Anweisung nicht vergessen
    print("a ist kleiner als 4")        # Da dieser Befehl Teil der Verzweigung ist, wird er eingerückt
else:                                   # Auch hier ist ein Doppelpunkt am Ende der Else-Anweisung
    print("a ist nicht kleiner als 4")  # Dieser Befehl muss auch eingerückt sein

print("Ende der Verzweigung")           # Da dieser Befehl nicht eingerückt ist, ist er nicht Teil der Verzweigung
````

````python
a=2.5

# Eine kompliziertere Verzweigung
if a == 1:
    print("a ist gleich 1")
elif a > 2 and a <= 4:
    print("a ist größer 2, aber kleiner oder gleich 4")
elif a == 2.5:
    print("a ist gleich 2.5")    
else:
    print("a ist entweder kleiner 1, oder größer als 4")
````

````python
# If mit mehreren Bedingungen (boolsche Verknüpfung besser als die Statements ineinander zu verschachteln)
if a > 0 and a < 5:
    print("a ist größer 0 und kleiner 5")
````

### 1.3. Iteration

Python kennt zwei Arten von Schleifen:
- ```while```-Schleife
- ```for```-Schleife

#### 1.3.1. ```while```-Schleife

Die ```while```-Schleife hat die allgemeine Form:

```Python
while <Ablaufbedingung B>:
    <Anweisungsblock A>
```

Auch hier gilt, dass die Befehle des Anweisungsblocks A im Vergleich zu den umliegenden Befehlen eingerückt sein müssen.

````python
i = 0
sum = 0

while i < 25:
    print('i',i)
    print('sum',sum)
    sum += i
    i += 1
    
print(sum)
````

#### 1.3.2. for-Schleife (Zählschleife)

Die ```for```-Schleife hat die allgemeine Form:

```Python
for <Zählervariable> in range([<start>,] <stop> [, <step>]):
    <Anweisungsblock A>
```

Die Parameter *start*, *stop* und *step* müssen Ganzzahlen sein, sie können auch negativ sein. Die Parameter *start* und *step* können weggelassen werden. Wenn *start* weggelassen wird, nimmt es den Wert 0 an, und wenn *step* weggelassen wird, nimmt es den Wert 1 an. Der Parameter *stop* ist exklusiv, das bedeutet, dass die *Zählervariable* niemals den Wert von *stop* annimmt.

Auch hier gilt, dass die Befehle des Anweisungsblock A im Vergleich zu den umliegenden Befehlen eingerückt sein müssen.

````python
# Hier hat range() nur einen Parameter, 
# i nimmt daher die Werte von 0 bis 4 an
for i in range(5):
    print("i =", i)
````

````python
# Hier hat range() zwei Parameter, start und stop
# i nimmt daher die Werte von 2 bis 6 an
for i in range(2, 7):
    print("i =", i)
````

````python
# Hier hat range() drei Parameter, start, stop und step
# i nimmt daher die Werte von 2, 5, 8, 11 an
for i in range(2, 12, 3):
    print("i =", i)
````

````python
# Schleifen können auch verschachtelt werden
for i in range(10):
    for j in range(10):
        print("i =", i, ", j =", j)
````

#### 1.3.3. ForEach-Schleife (Mengenschleife)

Die ForEach-Schleife hat die allgemeine Form:

```Python
for <Elementvariable> in <Menge>:
    <Anweisungsblock A>
```

Wer genau hinschaut, erkennt, dass die oben angeführte Zählschleife eigentlich eine Sonderform der Mengenschleife ist. Die Funktion *range()* generiert eine Menge, die die Zählwerte enthält.

Auch hier gilt, dass die Befehle des Anweisungsblock A im Vergleich zu den umliegenden Befehlen eingerückt sein müssen.

````python
# Hier generieren wir uns eine Menge mit unterschiedlichen Werten
m = [2, 1, 2.3, "Hallo", "Welt"]

for e in m:
    print("e =", e)
````

#### 1.3.4. Iterationsabbruch

Mit ```continue``` kann die aktuelle Iteration abgebrochen werden:

````python
for i in range(5):
    print("i =", i)
    if i > 1:
        continue    # Hier brechen wir die aktuelle Iteration ab, wenn i größer 1 ist
    print("Hallo")  # Dieser Befehl wird nicht ausgeführt, wenn i größer 1 ist
````

#### 1.3.5. Schleifenabbruch

Mit ```break``` kann die Ausführung der aktuellen Schleife abgebrochen werden:

````python
# Diese Schleife sollte eigentlich bis 9 zählen
for i in range(10):
    print("i =", i)
    if i == 3:
        break   # Da wir haber hier die Schleife abbrechen, sobald i gleich 3 ist, nimmt i keinen Wert größer als 3 an
````

````python
# Es lassen sich while-Schleifen auch mit dem boolschen Wert vorgegeben programmieren, wenn sichergestellt ist, dass die Abbruchbedingung erfüllt ist.

i = 0
while True: # Ohne erfüllte Abbruchbedingung ist while True eine Endlosschleife.
    print(i**2)
    if i**2 >= 25:
        break
    i += 1
````

## 2. Unterprogramme (Funktionen)

Funktionsdefinitionen in Python werden mit den Schlüsselwort **def** eingeleitet. 

Die allgemeine Form einer Funktionsdefinition schaut so aus:

```Python
def <funktionsname>([<parameterliste>]):
    <Anweisungsblock B>
```

Die Parameterliste kann 0 bis beliebig viele Eingabeparameter beinhalten. Die einzelnen Eingabeparameter werden durch einen Beistrich voneinander getrennt. Die Namen der Eingabeparameter müssen innerhalb der Liste eindeutig sein, und folgen denselben Regeln wie Variablennamen. Die Eingabeparameter werden innerhalb der Funktion als lokale Variablen behandelt.


Auch hier gilt, dass die Befehle des Anweisungsblock A im Vergleich zu den umliegenden Befehlen eingerückt sein müssen.

````python
# Hier definieren wir die Funktion f1
# Sie hat keinen einzigen Eingabeparameter
def f1():    # Auch hier darf der Doppelpunkt am Ende nicht fehlen
    print("f1() ausgeführt")
    print("f1() ist super")
````

````python
# Hier rufen wir f1 auf
f1()
f1()
````

````python
# Hier definieren wir die Funktion f2
# Sie hat einen Eingabeparameter x
def f2(a):
    print("f2( a ", a*2, ")")
````

````python
# Hier rufen wir f2 mit einer Zahl als Eingabeparameter auf
f2(441)
````

````python
# Und hier rufen wir foo2 mit einer Zeichenkette als Eingabeparameter auf
# Da Python dynamisch typisiert ist, ist das ohne weiteres möglich
f2("Hallo")
````

````python
# Hier definieren wir die Funktion f3
# Sie hat zwei Eingabeparameter x und y
def f3(x, y):
    print("f3(", x, ",", y*2, ")")
````

````python
# Hier rufen wir f3 mit einer Zahl und einer Zeichenkette als Eingabeparameter auf
f3(42, "bla")
````

Eine Funktion kann auch optionale Eingabeparameter haben. Diesen wird ein Initialwert zugewiesen.

````python
# Hier definieren wir die Funktion f4 mit optionalen Eingabeparametern
def f4(x, y = 5):
    print("f4(", x, ",", y, ")")
````

````python
# Hier rufen wir f4 mit einem Parameter auf. y bekommt den zugewiesenen Wert 5.
f4(1)

# Hier rufen wir f4 mit zwei Parametern auf.
f4(1,3)
````

Eine Funktion kann auch mithilfe des Befehls ``return`` einen oder mehrere Rückgabewerte zurückgeben. 

````python
# Die Funktion add() hat zwei Eingabeparameter a und b, 
# und gibt als Rückgabewert die Summe der Eingabeparameter zurück
def add(a, b):
    #print("inside add function",a+b)
    return a + b
````

````python
# Hier rufen wir add() mit 3 und 5 auf
add(3, 5)
````

Wenn ```return``` mehr als einen Rückgabewert hat, dann werden diese Rückgabewerte durch ein Komma voneinander getrennt. Mehrere Rückgabewerte können über den Zuweisungsoperator in mehrere Variablen gespeichert werden, die auch durch Kommas voneinander getrennt sind.

````python
# Die Funktion vertausche hat zwei Eingabeparameter, und gibt diese in umgekehrter Reihenfolge zurück
def vertausche(a, b):
    return b, a
````

````python
a = 3
b = 9

# Die Werte vor dem Vertauschen
print("a =", a)
print("b =", b)

print("Führe vertausche() aus")
a, b = vertausche(a, b)

# Die Werte nach dem Vertauschen
print("a =", a)
print("b =", b)
````

### 2.1. Docstrings

Funktionen (und andere Programmierstrukturen wie beispielsweise Klassen) werden normalerweise über Docstrings dokumentiert. Diese enthalten:
- eine Beschreibung, was die Funktion macht
- Eingabeparameter
- Ausgabeparameter

````python
def vertausche(a, b): # Festlegung des Docstrings als mehrzeiliger Kommentar nach Definition der Funktion.
    '''Vertauscht die Werte zweier Variablen.

    Inputs:
    a, b: Eingabeparameter
    Outputs:
    b, a: Vertauschte Werte'''
    return b, a
````

````python
# Hier wird auf den Docstring zugegriffen.

print(vertausche.__doc__)
````

````python
# Auch die Dokumentation von existenten Python-Funktionen lässt sich so aufrufen.
print(print.__doc__)
````

## 3. Beispiel: ggT

Mit allem, was wir in diesem Dokument gelernt haben, können wir ein Programm schreiben, welches den größten gemeinsamen Teiler zweier Zahlen berechnet:

````python
def vertausche(a, b):
    return b, a

def ggT(a, b):
    '''Rechnet den größter gemeinsamer Teiler aus.

    Inputs:
    a, b: Ganze Zahlen, deren größter gemeinsamer Teiler errechnet werden soll.
    Output:
    a: größter gemeinsamer Teiler der Eingabeparameter.'''

    if b > a: 
        a, b = vertausche(a, b)
        
    while b > 0:
        rest = a % b
        a = b
        b = rest
        print(rest)
    return a

print("ggT(130, 15) =", ggT(130, 15))
````
