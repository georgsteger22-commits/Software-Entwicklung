# 04 Rekursion und Iteration in Python
----

Rekursion funktioniert auch in Python so, dass eine Funktion sich selber aufruft, entweder direkt oder indirekt.

Schauen wir uns als Beispiel die Fakultätsfunktion an. Die Fakultät ist mathematisch definiert durch:

$
fac(n) :=
    \begin{cases}
        n\cdot fac(n-1) & \text{falls } n > 1 \hskip0.8cm \text{(Rekursionsschritt)}\\
        1                         & \text{sonst} \hskip1.9cm \text{(Rekursionsanfang)}
    \end{cases}
$

----

## 1. Lineare Rekursion

Eine linearrekursive Implementierung der Fakultät schaut in Python so aus:

```python
def fac_rekursiv(n):
    if n > 1:
        return n * fac_rekursiv(n-1)  # Rekursionsschritt
    else:
        return 1                      # Rekursionsanfang
```

```python
print(fac_rekursiv(5))
```

Schauen wir uns einmal die Ausführung der Faklutätsfunktion über den print Befehl an. Dafür verwenden wir eine debug-freundlichere Version, in der wir alle Zwischenwerte in lokale Variablen reinspeichern, damit man sie mit print auch ausgegeben werden kann.

```python
def fac_rekursiv_dbgfriendly(n):
    if n > 1:
        t1 = fac_rekursiv_dbgfriendly(n-1)
        t2 = n * t1
        print(t2)
        return t2
    else:
        print(1)
        return 1
```

```python
result = fac_rekursiv_dbgfriendly(5)
print("result =", result)
```

## 2. Endrekursion

Die Fakultätsfunktion kann auch endrekursiv implementiert werden. Die endrekursive Variante der Fakultät erwartet zwei Eingabeparameter, wobei beim ersten Aufruf der zweite Parameter immer 1 sein muss.

```python
def fac_endrekursiv(n, r):
    if n > 1:
        return fac_endrekursiv(n-1, r*n)  # Rekursionsschritt
    else:
        return r
```

```python
print(fac_endrekursiv(5, 1))
```

Hier muss natürlich der Programmierer aufpassen, dass er keinen falschen Wert als zweites Argument übergibt. Um die Angabe von falschen Eingabeparameter auszuschließen, gibt es zwei Möglichkeiten.

Die erste Möglichkeit ist die Verwendung von sog. [*Default-Werten* für Funktionsparameter](https://www.geeksforgeeks.org/default-arguments-in-python/), soweit das die jeweilige Programmiersprache unterstützt.

Die zweite Möglichkeit ist die Verwendung von zwei Funktionen, eine öffentliche und eine private:

Die öffentliche Funktion ist die, die der Programmierer aufruft, und die nur einen Eingabeparameter besitzt.
Diese öffentliche Funktion ruft dann die eigentliche Funktion, die von Programmierer nie direkt aufgerufen werden soll, mit den richtigen Eingabeparametern auf.

```python
# Die öffentliche Funktion, die der Programmierer aufruft
def fac_endrekursiv_public(n):
    return _fac_endrekursiv_intern(n, 1)

# Die private Funktion, die nie direkt aufgerufen werden soll
# Private Funktionen sind durch einen Unterstrich am Anfang gekennzeichnet
# Hier sieht man auch die Verwendung eines Default-Wertes für den zweiten Parameter "r"
def _fac_endrekursiv_intern(n, r = 1):
    if n > 1:
        return _fac_endrekursiv_intern(n-1, r*n)
    else:
        return r
```

```python
result = fac_endrekursiv_public(5)
print("result =", result)
```

```python
# Man könnte aber auch direkt die interne Fakultätsfunktion aufrufen und nur ein Funktionsargument übergeben.
# Es wird dann automatisch der bei der Funktionsdefinition angegebene Default-Wert verwendet
result = _fac_endrekursiv_intern(5)
print("result =", result)
```

## 3. Iteration vs. Rekursion

Die Fakultätsfunktion kann auch iterativ implementiert werden:

```python
def fac_iterativ(n):
    result = 1
    while n > 1:
        result = n * result
        n = n - 1
    return result
```

```python
result = fac_iterativ(5)
print("result =", result)
```

Wie Sie in der Vorlesung gehört haben, ist die rekursive Variante eines Algorithmus oft eleganter, die iterative aber effizienter. Schauen wir uns dies anhand eines Beispiels an.

Hier wird die Funktion iterativ und rekursiv ausgeführt. Auch wenn die Rechenzeiten von Ausführung zu Ausführung variieren, ist die iterative Rechenzeit geringer.

```python
import time

# Iterativ
start = time.perf_counter()
fac_iterativ(1000)
stop = time.perf_counter()
print("Iterative Rechenzeit:", stop-start, "ms")

#Rekursiv
start = time.perf_counter()
fac_rekursiv(1000)
stop = time.perf_counter()
print("Rekursive Rechenzeit:", stop-start, "ms")
```

Hier berechnen wir die Fakultät von 4000 mithilfe der iterativen Variante. Nachdem Sie auf Ausführen geklickt haben, bekommen Sie relativ schnell eine Antwort (die eine sehr große Zahl ist).

```python
print(fac_iterativ(4000))
```

Hier berechnen wir die Fakultät mithilfe der rekursiven Variante. Nachdem Sie auf Ausführen geklickt haben, rechnet der Computer viel länger als mit der iterativen Variante, und bricht dann mit einer Fehlermeldung ab. Die Fehlermeldung wird dadurch verursacht, dass aufgrund der vielen rekursiven Funktionsaufrufe irgendwann der Speicher am Stack ausgeht.

```python
print(fac_rekursiv(4000))
```

Versuchen wir nun, die maximale Anzahl an möglichen rekursiven Aufrufen der Fakultätsfunktion herauszufinden. Dazu verwenden wir eine globale Variable `depth`, in die wir die Anzahl der Aufrufe speichern.

```python
depth = 0

def fac_depth(n):
    
    # Das hier ist eine Variablendeklaration
    # Sie wird benötigt, damit der Python Interpreter weiß, dass depth schon als globale Variable existiert
    global depth
    
    depth = depth + 1
    
    if n > 1:
        return n * fac_depth(n-1)
    else:
        return 1

# Try-Except ist ein Mechanismus, um Fehlermeldungen (sog. Exceptions) abzufangen
try:
    result = fac_depth(4000)
    print("Fakultät erfolgreich berechnet")
except:
    print("Maximale Rekursionstiefe:", depth)
```

Wie wir sehen, liegt die maximale Rekursionstiefe bei ungefähr 3000. Um die maximale Rekursionstiefe zu erhöhen, müssten wir dem Stack mehr Speicher zuweisen.

### 4.1. Collatz Conjecture / Collatz Problem

Dies ist eines der Millennium Problem die das Clay Mathematics Institute im Jahre 2000 ausgegeben hat. Wir werden dieses Problem nicht lösen, aber wir werden es in iterativer und rekursiver Form darstellen.

$
collatz(r) :=
    \begin{cases}
        3 \cdot r + 1                     & \text{falls } (r \; mod \; 2) = 1 \\
        r/2                               & \text{sonst} 
    \end{cases}
$

```python
# Der Funktion collatzRek übergeben wir ausschließlich die Zahl r 
# Es wird kein Rückgabewert erwartet
def collatzRek(r):

    if r==4:
        print("Die Sequenz endet in einem Loop 4, 2, 1.")
    elif r%2==1:
        r=3*r+1
        print(r)
        collatzRek(r)
    else:
        r=r/2
        print(r)
        collatzRek(r)
    
    # Hier geben wir das Ergebnis zurück
    return None
        
# Hier rufen wir die Funktion auf für r=100
collatzRek(135168410665468406840654065416504)
```

Lasst uns nun eine iterative Funktion daraus schreiben:

```python
# Der Funktion collatzIter übergeben wir auch ausschließlich die Zahl r 
# Es wird kein Rückgabewert erwartet
def collatzIter(r):

    while r>4:
        if r%2==1:
            r=3*r+1
            print(r)
        
        else:
            r=r/2
            print(r)
                    
    print("Die Sequenz endet in einem Loop 4, 2, 1.")
    
    # Hier geben wir das Ergebnis zurück
    return None

# Hier rufen wir die Funktion auf für r=100
collatzIter(100)
```

### 4.2. Mittelwert berechnen

In diesem Beispiel berechnen wir den Mittelwert einer Verteilung die in Form eines Arrays vorliegt. Dabei starten wir mit dem iterativen Ansatz:

```python
# Hier wird die Verteilung in Form eines Array erstellt
array = [1, 2, 3, 4, 2, 3, 1, 2, 2, 3, 4, 5, 4, 2, 3, 4, 1, 2, 3, 2, 2, 3, 4, 5, 3, 2, 2, 3, 4, 5, 4, 2, 3, 2, 3, 4, 5]
```

```python
# Der Funktion mittelwertIter wird ein Array übergeben
# Es wird der Mittelwert des Arrays als Rückgabewert erwartet
def mittelwertIter(r):
    # Hier definieren wir eine Variable, die am Ende die Summe aller Array Elemente enthält
    sum=0
    
    for index in range(len(r)):
        sum=r[index]+sum

    # Hier geben wir das Ergebnis zurück
    return sum/len(r)
                
print("Der Rückgabewert ist", mittelwertIter(array))
```

Die selbe Funktion wird nun rekursiv entwickelt:

```python
# Hier wird die Verteilung in Form eines Array erstellt
array = [1, 2, 3, 4, 2, 3, 1, 2, 2, 3, 4, 5, 4, 2, 3, 4, 1, 2, 3, 2, 2, 3, 4, 5, 3, 2, 2, 3, 4, 5, 4, 2, 3, 2, 3, 4, 5]
```

```python
# Der Funktion mittelwertRek wird ein Array übergeben
# Es wird der Mittelwert des Arrays als Rückgabewert erwartet
def mittelwertRek(r,index=0):
    
    if index == len(r)-1:
        r[index]=r[index]/len(r)+r[index-1];
    elif index>=1:
        r[index]=r[index]/len(r)+r[index-1];
        mittelwertRek(r,index+1)
    else:
        r[index]=r[index]/len(r);
        mittelwertRek(r,index+1)
        return r[-1]

    
print("Der Rückgabewert ist", mittelwertRek(array))
```

### Aufgaben

Die Türme von Hanoi sind ein mathematisches Geduldsspiel, bei dem gelochte Scheiben auf drei Stäben liegen.
Zu Beginn liegen alle Scheiben auf Stab A, der Größe nach geordnet, mit der größten Scheibe unten und der kleinsten oben. Ziel des Spiels ist es, den kompletten Scheiben-Stapel von A nach C zu versetzen, ohne dass eine größere auf einer kleineren Scheibe liegt.

https://de.wikipedia.org/wiki/T%C3%BCrme_von_Hanoi

Die folgende rekursive Funktion errechnet die Mindestzahl um einen $n$ Scheiben hohen Stapel zu verschieben.
Schreiben Sie diese als Code.

$
hanoi(n) :=
    \begin{cases}
        2 \cdot hanoi(n-1) + 1 & \text{falls } n > 0\\
        0                         & \text{sonst}
    \end{cases}
$


```python
def hanoi(n):
    if n>0:
        return 2*hanoi(n-1)+1
    else:
        return 0
    
hanoi(7)
```

Formen Sie die Schleife zu einer rekursiven Funktion um.

Die programmierte Reihe der quadratischen Pyramidalzahlen gibt die benötigten Elemente, um eine Pyramide der Höhe $n$ zu bauen.

$\sum_{i=0}^n i^2 =  0^2 +1^2 + 2^2 + 3^2 + 4^2 +...$

```python
def sqpyr(n):
    i = 0
    r = 0
    while i <= n:
        r += i**2
        i += 1
    return r

for i in range(5):
    print(sqpyr(i))
```

```python
def sqpyr_rek(n):
    if n>=1:
        return n**2 + sqpyr_rek(n-1)
    else:
        return 0
    

for i in range(5):
    print(sqpyr_rek(i))
```

Man kann in rekursiven Funktionen Zwischenergebnisse anzeigen lassen. Was ändert sich, wenn die `print()`-Funktion vor oder nach dem rekursiven Funktionsaufruf steht? Warum?

```python
def summe(n, r = 0):
    #print(r)
    if n > 0:
        r_return = summe(n-1, r+n)  # Rekursionsschritt
        print(r)
    else:
        r_return = r
    return r_return

print("Summe:", summe(5))
```

### Aufgabe Eulersche Zahl
Entwerfen Sie einen Algorithmus, der die Eulersche Zahl $e$ näherungsweise berechnet. Nutzen Sie dabei die
folgende Formel:

\begin{equation}e = \sum_{i=0}^{n} \frac{1}{i!} = 1 + 1 + \frac{1}{2} + \frac{1}{2\cdot3} + \frac{1}{2\cdot3\cdot4} + \frac{1}{2\cdot3\cdot4\cdot5}+ ...\end{equation}

Als Eingabe erhält der Algorithmus die natürliche Zahl n, die die Anzahl der Summanden angibt.
1. Schreiben Sie die Funktion `e_iterative(n)`, welche iterativ die Eulersche Zahl $e$ unter Verwendung von `fac_rekursiv(n)` aus Abschnitt 1 berechnet und zurückgibt.
2. Schreiben Sie eine Methode `e_recursive(n)`, die rekursiv die Eulersche Zahl $e$ unter Verwendung von `fac_rekursiv(n)` aus Abschnitt 1 berechnet und zurückgibt.

```python
def e_iterative(n):
    sum=0
    for i in range(n+1):
        sum+=1/fac_iterativ(i)
    return sum

print(e_iterative(5))

def e_recursive(n):
    if n>=0:
        return 1/fac_iterativ(n)+e_recursive(n-1)
    else:
        return 0

print(e_recursive(5))    
```
