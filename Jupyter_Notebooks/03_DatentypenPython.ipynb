## 1. Typbestimmung zur Laufzeit

Python ist eine sogenannte *dynamisch typisierte* Programmiersprache. Das heißt, dass die Datentypen von Variablen dynamisch anhand vom Inhalt der Variablen bestimmt werden, und Variablen ihren Datentypen auch jederzeit ändern können.

Wenn man dynamisch typisierte Sprachen verwendet, kann es vorkommen, dass ein bestimmter Datentyp vorrausgesetzt wird. Um den Datentypen einer Variablen zu bestimmen, steht in Python die `type()` Funktion zur Verfügung. Diese Funktion erwartet einen Eingabeparameter, und gibt den Datentypen des Eingabeparameters zurück.

Beispiel für die Verwendung der `type()`-Funktion:

```python
# Diese Funktion addiert 2 Ganzzahlen
# Da Python dynamisch typisiert ist, müssen wir sicherstellen, dass wirklich 2 Ganzzahlen übergeben wurden
def add(a, b):
    # Dieses IF-Statement überprüft die Datentypen der Eingabeparameter
    if type(a) != int or type(b) != int:
        print("Fehler: Mindestens ein Eingabeparameter hat den falschen Datentypen")
    else:
        return a + b
```

```python
# Hier rufen wir dir obrige Funktion mit den richtigen Datentypen auf
print(add(1, 3))
```

```python
# Hier hat der erste Eingabeparameter den falschen Datentypen
print(add("1", 3))
```

```python
# Eine Gleitkommazahl ist keine Ganzzahl, daher geben wir hier eine Fehlermeldung aus
print(add(3, 3.5))
```

## 2. Der "Nichts" Datentyp

Es gibt immer wieder Situationen, wo ein Eingabeparameter (vielleicht weil er optional ist) oder ein Rückgabewert (vielleicht weil ein Fehler aufgetreten ist) den Wert "Nichts" annehmen soll. In Python gibt es dafür den speziellen Datentypen `NoneType`, dessen Wertebereich genau einen Wert enthält, nämlich `None`.

`None` repräsentiert die Abwesenheit eines Wertes. Dass die "Abwesenheit eines Wertes" durch einen speziellen Wert repräsentiert wird (was ja eigentlich ein Widerspruch in sich ist), hat damit zu tun, dass der Computer mit "Nichts" anders nicht umgehen kann. Der Computer versteht nur Bitfolgen, und deshalb muss "Nichts" durch eine spezielle Bitfolge dargestellt werden. Und in Python heißt eben diese spezielle Bitfolge `None`.

```python
print(type(None))
```

Beispiele für die Verwendung von `None`:

```python
# Wir können die obrige add() Funktion so umschreiben, dass sie statt einer Fehlermeldung "None" zurückgibt
def add(a, b):
    # Dieses IF-Statement überprüft die Datentypen der Eingabeparameter
    if type(a) != int or type(b) != int:
        return None
    else:
        return a + b
```

```python
# Dann kann man den Rückgabewert auf None testen, um herauszufinden, ob ein Fehler aufgetreten ist
# In diesen Beispiel hat der Funktionsaufruf fehlerfrei funktioniert
result = add(2, 3)
if result == None:
    print("Ein Fehler ist aufgetreten")
else:
    print("Das Ergebnis ist", result)
```

```python
# In diesen Beispiel hat der Funktionsaufruf nicht fehlerfrei funktioniert
result = add("2", 3)
if result == None:
    print("Ein Fehler ist aufgetreten")
else:
    print("Das Ergebnis ist", result)
```

```python
# Man kann auch "None" für optionale Eingabeparameter verwenden.
# In dieser Funktion ist b ein optionaler Eingabeparameter.
# Wenn bei einem Funktionsaufruf dieser optionale Eingabeparameter nicht verwendet wird,
# dann nimmt b den Standardwert None an.
def f(a, b = None):
    if b == None:
        print("Der optionale Eingabeparameter b wird nicht verwendet")
    else:
        print("Der optionale Eingabeparameter b wird verwendet")
```

```python
# Hier verwenden wir den optionalen Eingabeparameter b
f(3, 4)
```

```python
# Hier verwenden wir den optionalen Eingabeparameter b nicht. b nimmt den Standardwert None an.
f(3)
```

```python
# Wir können "None" auch verwenden, um Variablen anzulegen (zur Erinnerung, eine Variable wird automatisch 
# bei der ersten Zuweisung angelegt), bevor wir einen tatsächlichen Wert haben, 
# den wir in dieser Variable speichern wollen

# Hiermit legen wir die Variable x an
x = None
# Hier sehen wir, dass die Variable x existiert, aber keinen Wert enthält ("None" repräsentiert ja "Nichts")
print("x =", x)
print("Datentyp:", type(x))
```

```python
# Eine Funktion ohne Ausgabeparameter gibt None zurück
def multiply(a,b):
    c = a*b
    print("Ergebnis:", c)


# Hier wird die Funktion f ausgeführt
d = multiply(3,5)
print("d =", d)
```

## 3. Elementare Datentypen

### 3.1. Zahlen

#### 3.1.1. Ganzzahlen

Jede Zahl ohne Nachkommastellen wird von Python automatisch als Ganzzahl interpretiert. Ganzzahlen haben in Python einen theoretisch unbegrenzten Wertebereich, in der Praxis ist dieser vom verfügbaren Arbeitsspeicher begrenzt.

In Python können für Ganzzahlen alle arithmetischen und Vergleichsoperationen verwendet werden.

Beispiele:

```python
# a wird eine Ganzzahl zugewiesen
a = 12345
print(type(a))
```

```python
# Eine Addition mit Ganzzahlen
b = 34 + 56
print("34 + 56 =", b)
print(type(b))
```

#### 3.1.2. Gleitkommazahlen

Jede Zahl mit Nachkommastellen wird von Python automatisch als Gleitkommazahl interpretiert. Python verwendet für Gleitkommazahlen 64-bit Zahlen im IEEE 754 Format, d.h. der Wertebereich von Gleitkommazahlen geht von $\approx 2.23\cdot 10^{-308}$ bis $\approx 1.78\cdot 10^{308}$.

In Python können für Gleitkommazahl alle arithmentischen und Vergleichsoperationen verwendet werden.

Beispiele:

```python
# a wird eine Gleitkommazahl zugewiesen
a = 123.45
print(type(a))
```

```python
# Eine Addition mit Gleitkommazahlen
# 34.3 + 56.4 ergibt normalerweise 90.7
# Aufgrund von Darstellungsfehlern bei Gleitkommazahlen ist das Ergebnis hier aber 90.69999999999999
b = 34.3 + 56.4
print("34.3 + 56.4 =", b)
print(type(b))
```

```python
# Wenn man eine Zahl ohne Nachkommastelle als Gleitkommazahl speichern will, hängt man einfach '.' hinten dran
c = 434.
print("c =", c)
print(type(c))
```

```python
# Statt der Normalen Schreibweise kann man auch die wissenschaftliche Schreibweise verwenden
# 'E' bzw. 'e' steht für "Zehnerpotenz", d.h. 4.34E2 = 4.43*10^2 = 434
d = 4.34E2
print("d =", d)
print(type(d))
```

#### 3.1.3. Komplexe Zahlen

Python hat sogar einen eigenen Datentypen für komplexe Zahlen. Komplexe Zahlen werden in der Form `3+5j` geschrieben, dabei kennzeichnet 'j' bzw. 'J' den komplexen Teil.

In Python können für komplexe Zahlen alle arithmentischen und Vergleichsoperationen verwendet werden.

Beispiele:

```python
# a wird eine komplexe Zahl zugewiesen
R=100
Xl=10
Xc=-5

Z = (R+Xl*1j+Xc*1j)
print("a =", Z)
print(type(Z))
```

```python
# Komplexe Zahlen können natürlich auch Nachkommastellen haben
a = 3.6+4.4j
print("a =", a)
print(type(a))
```

```python
# Eine Addition mit komplexen Zahlen
b = (3+4j) + (5+3j)
print("(3+4j) + (5+3j)=", b)
print(type(b))
```

```python
# Vergleiche sind bei complexen Zahlen in zwei Kategorien zu unterteilen
# Einmal gibt es die Vergleiche zur (Un)-Gleichheit
# Dabei wird sowohl Realteil als auch Imaginärteil auf (Un)-Gleichheit geprüft

a = 3+4j
b = 3+4j
if a==b:
    print(b,"ist gleich", a)
else:
    print(b,"ist nicht gleich", a)
    
a = 3+5j
b = 3+4j
if a==b:
    print(b,"ist gleich", a)
else:
    print(b,"ist nicht gleich", a)
    
a = 3+4j
b = 2+4j
if a==b:
    print(b,"ist gleich", a)
else:
    print(b,"ist nicht gleich", a)
```

```python
# Die zweite Kategorie bei Vergleichen komplexer Zahlen entspricht <,>,>=,<=
# Diese Verleiche sind in python nicht direkt unterstützt sobald ein complexer Datentyp involviert ist.
# Dies wird auch in einer Error Message entsprechend aufgezeigt.
a = 3+5j
b = 3+4j
if a<b:
    print(a,"ist kleiner als", b)
else:
    print(a,"ist nicht kleiner als", b)
```

```python
# Um die Zahlen zu vergleichen müsste man entsprechend den Betrag oder den Realteil zur Hand nehmen.

a = 3-5j
print("Betrag:", abs(a))
print("Realteil:", a.real)
print("Imaginärteil:", a.imag)
```

#### 3.1.4. Implizite Casts bei Zahlen

In Python finden implizite Casts nur von Ganzzahlen zu Gleitkommazahlen statt. 

Wenn auch nur ein Operand einer arithmetischen Operation eine Gleitkommazahl ist, werden automatisch alle Operanden in Gleitkommazahlen konvertiert und das Ergebnis als Gleitkommazahl zurückgegeben. Dies passiert, um Genauigkeitsverluste zu verhindern (alle Ganzzahlen sind als Gleitkommazahl darstellbar, aber nicht alle Gleitkommazahlen als Ganzzahlen).

Bei der Division werden auch automatisch alle Operanden (auch wenn beide Operanden Ganzzahlen sind) in Gleitkommazahlen umgewandelt, und das Ergebnis als Gleitkommazahl zurückgegeben. Auch hier will man Genauigkeitsverluste vorbeugen. Bei der Ganzzahl-Division '//' passiert dies jedoch nicht (Darum heißt sie ja auch Ganzzahl-Division).

Generell kann man sagen, wenn Genauigkeitsverluste drohen, werden Ganzzahlen automatisch in Gleitkommazahlen umgewandelt (mit der Ausnahme der Ganzzahl-Division).

Beispiele:

```python
# Addition einer Ganzzzahl und einer Gleitkommazahl
a = 4 + 45.3
print("a =", a)
print(type(a))
```

```python
# Division zweier Ganzzzahlen
b = 6 / 3
print("b =", b)
print(type(b))
```

```python
# Komplexerer Arithmetischer Ausdruck, der eine Gleitkommazahl enthält
c = (2+4)*5 - (34+2)*3 + 4.4
print("c =", c)
print(type(c))
```

#### 3.1.4. Explizite Casts bei Zahlen

Mit der Funktion `int()` kann jeder Wert, sofern überhaupt möglich, explizit in eine Ganzzahl gecastet werden. Als Eingabeparameter wird dabei der zu castende Wert übergeben.

Mit der Funktion `float()` kann jeder Wert, sofern überhaupt möglich, explizit in eine Gleitkommazahl gecastet werden. Als Eingabeparameter wird dabei der zu castende Wert übergeben.

Mit der Funktion `complex()` kann jeder Wert, sofern überhaupt möglich, explizit in eine komplexe Zahl gecastet werden. Als Eingabeparameter wird dabei der zu castende Wert übergeben.

Beispiele:

```python
# Expliziter Cast Gleitkommazahl => Ganzzahl
a = int(34.9)
print("a =", a)
print(type(a))
```

```python
# Expliziter Cast String => Gleitkommazahl
a = int("64")
print("a =", a)
print(type(a))
```

```python
# Expliziter Cast Ganzzahl => Gleitkommazahl
c = float(34)
print("c =", c)
print(type(c))
```

```python
# Expliziter Cast "String" => Gleitkommazahl
d = float("64.5")
print("d =", d)
print(type(d))
```

```python
# Das Dezimalzeichen ist mit einem Punkt zu wählen
# Python unterstützt keinen Beistrich/Komma 
d = float("64,5")
print("d =", d)
print(type(d))
```

```python
# Expliziter Cast Ganzzahl => komplexe Zahl
e = complex(34)
print("e =", e)
print(type(e))
```

```python
# Expliziter Cast String => komplexe Zahl
f = complex("4+6j")
print("f =", f)
print(type(f))
```

### 3.2. Wahrheitswerte

Python kennt auch einen Datentypen für Wahrheitswerte, den Datentyp `bool`. Dessen Wertebereich enthält genau zwei Werte: `True` und `False` (auf Groß- und Kleinschreibung achten!).

In Python können für Wahrheitswerte alle logischen und Vergleichsoperationen verwendet werden.

Python verwendet intern Zahlen, um Wahrheitswerte zu speichern. dabei wird `True` 1 und `False` 0 zugewiesen. Aus diesem Grund können auch arithmetische Operationen mit Wahrheitswerten verwendet werden.

Mit der Funktion `bool()` kann jeder Wert, sofern überhaupt möglich, explizit in einen Wahrheitswert gecastet werden. Als Eingabeparameter wird dabei der zu castende Wert übergeben.

Beispiele:

```python
# a wird ein Wahrheitswert zugewiesen
a = True
print("a =", a)
print(type(a))
```

```python
# Und Verknüpfung
b = True and False
print("True and False =", b)
print(type(b))
```

```python
# Eine Addition mit Wahrheitswerten
# Hier findet ein impliziter Cast zu int statt
c = True + True
print("True + True =", c)
print(type(c))
```

```python
# Expliziter Cast Ganzzahl => Wahrheitswert (0 wird zu False, alles andere zu True)
d = bool(1)
print("d =", d)
print(type(d))
```

```python
# Expliziter Cast String => Wahrheitswert ("" d.h leerer String wird zu False, alles andere zu True)
e = bool("")
print("e =", e)
print(type(e))
```

#### 3.2.1. Kurzschlussauswertung (short-circuit evaluation)

Bei logischen Ausdrücken wird die Auswertung des Ausdrucks vorzeitig abgebrochen, sobald das Ergebnis schon feststeht.

Zum Beispiel, wenn bei einer Und-Verknüpfung der erste Operand `False` ist, ist der gesamte Ausdruck `False`, unabhängig vom Wert des zweiten Operanden. Daher kann man sich die Auswertung des zweiten Operanden sparen. 
Beispiel hierzu:

```python
# Hier definieren wir uns eine Funktion, die unseren ersten Operanden repräsentiert, und False zurückgibt
def firstOperand():
    print("Evaluating first operand")
    return True

# Hier definieren wir uns eine Funktion, die unseren zweiten Operanden repräsentiert, und True zurückgibt
def secondOperand():
    print("Evaluating second operand")
    return True

# Hier ist unsere Und-Verknüpfung
# Wie man an der Ausgabe unten erkennen kann, wird der zweite Operand gar nicht ausgewertet.
# Wenn man nun den ersten Operanden True zurückgeben lässt, dann erst wird der zweite Operand ausgewertet
print(firstOperand() or secondOperand())
```

Dies kann man für Checks ausnutzen (d.h. man kann sich If-Abfragen einsparen).

Hier ein Beispiel, wie man die Kurzschlussauswertung verwenden kann, um eine Division durch 0 zu verhinden:

```python
# Hier definieren wir uns ein paar Variablen, die wir im Folgenden verwenden
a = 4
b = 0
```

```python
# Hier haben wir eine If-Bedingung, wo wir a durch b dividieren
# Und zufälligerweise ist b gleich 0, was einen "division by zero" Fehler ergibt
if a / b == 2:
    print("a/b==2")
else:
    print("a/b!=2")
```

```python
# Nun können wir uns eine zweite If-Abfrage hinzufügen, die diesen "division by zero" Fehler abfängt
if b != 0:
    if a / b == 2:
        print("a/b==2")
    else:
        print("a/b!=2")
else:
    print("a/b!=2")
```

```python
# Oder wir können dasselbe erreichen, indem wir die Kurzschlussauswertung ausnutzen
# Wenn 'b != 0' False ist, wird a / b == 2 gar nicht mehr ausgewertet, und daher kann auch kein Fehler mehr auftreten
if b != 0 and a / b == 2:
    print("a/b==2")
else:
    print("a/b!=2")
```

## 3.3. Einzelne Zeichen

Python kennt keinen eigenen Datentypen für "einzelnes Zeichen". In Python ist ein einzelnes Zeichen ein String der Länge 1.

Mit der Funktion `ord()` kann der Zahlenwert (laut Codetabelle) eines einzelnen Zeichens ermittelt werden. Als Eingabeparameter wird dabei ein String der Länge 1 übergeben.

Mit der Funktion `chr()` kann ein Zahlenwert (laut Codetabelle) in ein einzelnen Zeichens, d.h. einen String der Länge 1, konvertiert werden. Als Eingabeparameter wird dabei eine Ganzzahl übergeben.

Beispiele:

```python
# a wird ein einzelnes Zeichen (String der Länge 1) zugewiesen
a = "h"
print("a =", a)
print(type(a))
```

```python
# Hier wird der Zahlenwert des Zeichens "A" ermittelt
b = ord("h")
print("b =", b)
print(type(b))
```

```python
# Hier wird ein Zahlenwert in ein Zeichen konvertiert
c = chr(104)
print("c =", c)
print(type(c))
```

```python
# Hier eine Methode, um ein kleines "a" in ein großes "A" umzuwandeln
d = ord("a") # Zuerst holen wir uns den Zahlenwert von "a"
d = d - 32 # Dann subtrahieren wir 32 (das funktioniert nicht nur für a, sondern für alle Kleinbuchstaben)
e = chr(d) # Dann wandeln wir den neuen Zahlenwert in ein Zeichen um
print("e =", e) # Voila, e enthält "A"
print(type(e))
```

### 4.1. Zeichenketten (Strings)

In Python sind Zeichenketten eine beliebig lange Auflistung von einzelnen Unicode-Zeichen.

Zeichenketten werden durch `"` bzw. `'` eingefasst. Mehrzeilige Zeichenketten werden durch `"""` eingefasst.

Mit dem `+` Operator können mehere Strings aneinandergehängt werden.

Die Funktion `len()` gibt die Länge des übergebenen Strings zurück.

```python
# a wird ein einzeiliger String zugewiesen
a = "Hallo"
print("a =", a)
print(type(a))
```

```python
# Hier fügen wir mehrere Strings zu einem zusammen
c = "Hallo" + " " + "Welt"
print("c =", c)
print(type(c))
```

```python
# Hier ermitteln wir die Anzahl der Zeichen in einem String
d = len("Hallo Welt")
print("d =", d)
print(type(d))
```

```python
# Hier greifen wir auf das 3. Zeichen des Strings zu
d = "Hallo Welt"
e = d[2] # Index beginnt bei 0, deshalb hat das 3. Zeichen den Index 2
print("e =", e)
print(type(e))
```

```python
# Hier wandeln wir eine Zahl in einen String um
f = str(123)
print("f =", f)
print(type(f))
```

```python
# Hier wandeln wir einen Wahrheitswert in einen String um
g = str(True)
print("g =", g)
print(type(g))
```

### 4.2. Arrays

Python kennt keinen Datentypen für statische Arrays. Stattdessen werden dynamische Arrays (Listen) verwendet.

### 4.3. Listen (dynamische Arrays)

In Python stellt der Datentyp `list` dynamische Arrays bereit. Da Python dynamisch typisiert ist, können die Listenelemente beliebige Datentypen annehmen.

Eine Liste kann in Python erzeugt werden, indem man die einzelnen Elemente, die durch Beistriche getrennt werden, mit eckigen Klammern `[]` umgibt:

```python
# Hier erzeugen wir eine Liste und speichern sie in a
a = [1, 2, 4]
print("a =", a)
print(type(a))
```

```python
# So können wir eine leere Liste erzeugen
a = []
print("a =", a)
print(type(a))
```

Mit dem Postfix-Operator [] kann auf die einzelnen Elemente zugegriffen werden. Als Argument wird hierbei der Index des Zeichens übergeben (der Index startet wieder bei 0):

```python
# Hier erzeugen wir eine Liste, damit wir was zum zugreifen haben
a = [1, 2, 3, 4]
# Hier greifen wir auf das 2. Element zu
b = a[1]
print("b =", b)
print(type(b))
```

```python
# Man kann natürlich auch eine Variable verwenden für den Index verwenden
# Hier verwenden wir die Zählervariable i einer for-Schleife, um die einzelnen Elemente der Liste auszugeben
for i in range(4):
    print("Element mit Index", i, ":", a[i])
```

```python
# Den Operator [] kann man auch verwenden, um Elemente an einem bestimmten Index zu speichern.
# Das vorher dort gespeicherte Element wird dabei ersetzt.
# Hier speichern wir "a" an Stelle 2.
a[2] = "my new element"
print("a =", a)
```

Mit der Funktion `append()` können neue Elemente an die Liste angehängt werden. Diese Funktion ist eine Klassenmethode (näheres zu Klassenmethoden in den weiterführenden Lehrveranstaltungen), und wird daher in der Form `<ListenName>.append(...)` aufgerufen:

```python
# Erzeugen wir uns eine Liste myList
myList = [1, 2, 3, 4]
# Geben diese Liste aus
print("myList =", myList)
# Hängen nun "a" an die Liste an
myList.append("a")
# Und geben die Liste erneut aus
print("myList =", myList)
```

Mit dem Befehl `del` können Elemente aus der Liste gelöscht werden. Dabei wird der Index des zu löschenden Elements mithilfe des Postfix-Operators `[]` angegeben: 

```python
# Geben wir die Liste myList aus
print("myList =", myList)
# Löschen wir nun das Element an der 3. Stelle aus der Liste myList
del myList[2]
# Und geben die Liste erneut aus
print("myList =", myList)
```

`pop` entfernt das letzte Element.

```python
# Geben wir die Liste myList aus
print("myList =", myList)
# Löschen wir nun das Element an der letzten Stelle aus der Liste myList
myList.pop()
# Und geben die Liste erneut aus
print("myList =", myList)
```

Die Funktion `len()` gibt die Anzahl der in der Liste gespeicherten Elemente zurück:

```python
# Hier ermitteln wir die Anzahl der Elemente von myList
b = len(myList)
print("b =", b)
print(type(b))
```

Listen können auch geschachtelt werden. Solche verschachtelten Listen kann man verwenden, um *mehrdimensionale Arrays* zu repräsentieren:

```python
c = [[2,3], [4, 5], [6, 7, 8]]
print("c =", c)
print(type(c))
```

Um auf die einzelnen Elemente von verschachtelten Listen zuzugreifen, hängt man einfach mehrere Postfix-Operatoren `[]` hintereinander:

### 4.4. Maps (assoziative Arrays)

In Python stellt der Datentyp `dict` assoziativen Arrays bereit. Da Python dynamisch typisiert ist, können die Elemente eines assoziativen Arrays beliebige Datentypen annehmen.

Eine Map kann in Python erzeugt werden, indem man die einzelnen Key-Value Paare, die durch Beistriche getrennt werden, mit geschwungenen Klammern `{}` umgibt. Bei den Key-Value Paaren wird der Key durch einen Doppelpunkt `:` vom Value getrennt.


```python
# Hier legen wir eine Map mit drei Key-Value Paaren an
a = {"Baujahr": 2012, "Leistung": 75, "Hersteller": "Vovlo"}
print("a =", a)
print(type(a))
```

```python
# So können wir eine leere Map erzeugen
a = {}
print("a =", a)
print(type(a))
```

Mit dem Postfix-Operator `[]` kann man wieder auf die einzelnen Elemente zugreifen Als Argument wird jedoch statt eines Index der Key des Elements übergeben:

```python
# Hier erzeugen wir ein assoziatives Array mit Inhalt, damit wir was zum zugreifen haben
a = {"key1": 1, "key2": "somestring", "key3": [1, 2, 3]}
# Hier greifen wir auf das Element mit dem Key "key2" zu
b = a["key2"]
print("b =", b)
print(type(b))
```

```python
# Den Operator [] kann man auch verwenden, um Elemente mit einem bestimmten Key zu speichern.
# Falls schon ein Element mit denselben key existiert, wird es dabei ersetzt.
# Hier speichern wir "a" unter dem key "key4".
a["key4"] = "a"
print("a =", a)
```

Mit dem Befehl `del` können Elemente aus der Map gelöscht werden. Dabei wird der Key des zu löschenden Elements mithilfe des Postfix-Operators `[]` angegeben:

```python
# Geben wir die Map a aus
print("a =", a)
# Löschen wir nun das Element mit dem Key "key2"
del a["key2"]
# Und geben die Map erneut aus
print("a =", a)
```

Die Funktion `len()` gibt die Anzahl der in der Map gespeicherten Elemente zurück:

```python
# Hier ermitteln wir die Anzahl der Elemente von a
b = len(a)
print("b =", b)
print(type(b))
```

### 5.1. Finde die größte Zahl in einem Array

Lasst uns eine Funktion schreiben, welche die größte Zahl in einem Array findet:

```python
# Hier definieren wir unser Array, das 8 Zahlen enthält
array = [4, 76, 23, 77, 34, 65, 45, 67]

# Dieser Funktion übergeben wir das Array und die Anzahl der Elemente im Array als Eingabeparameter,
# und als Rückgabewert erhalten wir die größte Zahl im Array
def maxNumberInArray(array, size):
    # Hier definieren wir eine Variable, die am Ende die größte Zahl enthält
    result = None  
    # Hier iterieren wir über alle Elemente im Array
    for i in range(size):
        # Das hier könnten wir uns sparen, es steht hier nur, damit wir das aktuelle Element im print ausgeben
        e = array[i]
        print(e)
        # Hier überprüfen wir, ob das aktuelle Element größer als das derzeit in result gespeicherte ist
        # Wir verwenden auch hier die Kurzschlussauswertung, da der Operator > für None nicht definiert ist
        if result == None or e > result:
            result = e
            print("Neuer Spitzenwert ist",result)
            
    # Hier geben wir das Ergebnis zurück
    return result
        
# Hier rufen wir die Funktion für unser Array auf
print("Die größte Zahl im Array ist", maxNumberInArray(array, len(array)))
```

### 5.2. Fakultät

Die [Fakultät](https://de.wikipedia.org/wiki/Fakult%C3%A4t_(Mathematik)) ist in der Mathematik eine Funktion, die einer natürlichen Zahl das Produkt aller natürlichen Zahlen (ohne Null) kleiner und gleich dieser Zahl zuordnet.

D.h. die Fakultät von $4$, geschrieben $4!$, ist $4! = 4 \cdot 3 \cdot 2 \cdot 1 = 24$.

Lasst uns eine Funktion schreiben, welche die Fakultät berechnet:

```python
# Diese Funktion berechnet die Fakultät von n
def fakultaet(n):
    # Hier erzeugen wir unsere Ergebnisvariable und initialisieren sie mit 1
    result = 1
    # Hier bilden wir das Produkt aller Zahlen <= n
    # In der range-Funktion schreiben wir n+1, weil der stop-wert von range exklusive ist
    for i in range(1, n+1):
        result = result * i
        print(i,result)
    # Hier geben wir das Ergebnis zurück
    return result

print("5! =", fakultaet(5))
```

### 5.3. Bubblesort

Eine häufige Operation auf Arrays ist das Sorten eines Arrays. Bubblesort stellt dabei den einfachsten Sortieralgorithmus dar. Er ist allerdings nicht sehr effizient, und wird daher in der Praxis kaum eingesetzt.

Für eine genauere Beschreibung von Bubblesort siehe [hier](https://de.wikipedia.org/wiki/Bubblesort) und [hier](http://interactivepython.org/runestone/static/pythonds/SortSearch/TheBubbleSort.html).

Schauen wir uns im folgenden die Ausführung von Bubblesort Schritt für Schritt an:

```python
# Diese Funktion implementiert Bubblesort
def bubbleSort(a):
    # Die Anzahl unser Durchgänge ist gleich Anzahl der Elemente -1
    for i in range(len(a)-1):
        # Hier gehen wir das Array von links nach rechts durch
        # Aus Effizienzgründen gehen wir nie bis ganz ans Ende, und mit jedem Durchgang um eins weniger nach rechts
        print(a)
        for j in range(len(a)-1):
            # Hier überprüfen wir, ob das erste Element größer als das zweite Element ist
            if a[j] > a[j+1]:
                # Wenn das der Fall ist, vertauschen wir hier die Elemente
                a[j], a[j+1] = a[j+1], a[j]
    # Hier geben wir das sortierte Array zurück
    return a


# Hier führen wir Bubblesort aus
result = bubbleSort([54, 26, 93, 17, 77, 31])
# Und geben das Ergebnis aus
print(result)
```

### 5.4. Primzahlen

Hier definieren wir uns eine Funktion, die testet, ob eine Ganzzahl eine Primzahl ist:

```python
# Eine Primzahl ist eine Zahl, die nur durch 1 und sich selbst ohne Rest teilbar ist. 
# Daher dividieren wir die Zahl n durch alle Zahlen zwischen 2 und n/2, 
# und überprüfen, ob es einen Rest gibt oder nicht.

def is_prime(n):
    for i in range(2, n//2+1):
        # Das hier könnten wir uns sparen, es steht hier nur, damit wir den aktuellen Rest im Debugger sehen
        rest = n % i
        if rest == 0:
            return False
    return True
```

```python
# Ist 4 eine Primzahl?
print(is_prime(4))
```

```python
# Ist 7 eine Primzahl?
print(is_prime(7))
```

```python
# In diesem Beispiel bestimmen wir alle Primzahlen zwischen 2 und 500

# Wir definieren uns eine leere Liste, wo wir alle gefundenen Primzahlen reinschreiben
primzahlen = []

# Hier gehen wir alle Zahlen von 2 bis 500 durch
for i in range(2, 501):
    # Wenn i eine Primzahl ist, dann hänge i an die Liste an
    if is_prime(i):
        primzahlen.append(i)

print("Primzahlen von 1 bis 500:", primzahlen)
```
