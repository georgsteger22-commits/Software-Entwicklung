# Zusatzaufgabe: ISBN

Jedes Buch dieser Welt ist durch eine 13-stellige Internationale Standardbuchnummer (ISBN-13) eindeutig
gekennzeichnet. Die Letzte Ziffer der ISBN ist eine sogenannte Prüfziffer, die entscheidet, ob es sich wirklich
um eine gültige ISBN handelt, oder ob z.B. ein Tippfehler oder Zahlendreher vorliegt. Diese berechnet sich aus
den ersten 12 Ziffern der ISBN wie folgt:
- Multipliziere jede zweite Ziffer (2-te, 4-te, 6-te ...) mit 3.
- Summiere anschließend die Ergebnisse aller zwölf Ziffern auf.
- Berechne dann den Abstand des Ergebnisses zum nächst höheren Vielfachen von 10.
- Das Ergebnis entspricht der Prüfziffer.

[Beispiel zur Berechnung](https://de.wikipedia.org/wiki/Internationale_Standardbuchnummer#Formeln_zur_Berechnung_der_Pr%C3%BCfziffer)

1. Schreiben Sie eine Funktion zur Validierung von ISBN-13 Nummern. Das Programm erhält als Eingabe-Argument eine 13-stellige Ziffernfolge (als String aus Zahlen, ohne '-') und soll für diese prüfen, ob es sich dabei um eine gültige ISBN-13 Nummer handelt.
2. Modifizieren Sie Ihr Programm nun so, dass nicht nur eine, sondern eine Liste von Ziffernfolgen als Eingabe-Argumente angegeben werden können.
3. Versuchen Sie die folgenden Eingabeargumente:\
9780060839871 9783127323204 9780007231607 9780261102385 9708261103285
