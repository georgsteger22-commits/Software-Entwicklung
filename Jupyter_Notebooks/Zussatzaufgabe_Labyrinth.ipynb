### Aufgabe
Schreiben sie eine rekursive Funktion `escape`, die automatisch einen Weg vom Startpunkt zum Endpunkt (beide gegeben) berechnet. Beachten Sie, dass die Farbe der Wände Schwarz (0) und die Farbe des Weges Weiß (255) ist. Speichern Sie den Weg in einen neuen Array mit Nullen als Wert 1 ab.

Speichern Sie zwei Bilder ab: Auf einem Bild soll der Weg eingezeichnet werden, den der Algorithmus geht, inklusive aller Sackgassen. In das zweite Bild soll nur der korrekte Weg eingezeichnet werden (ohne unnötige Umwege!). In Abbildung 1 ist eine beispielhafte Ausgabe gegeben.

Die Array-Koordinaten des Ein- und des Ausganges sind:
- Eingang: 53, 0
- Ausgang: 0, 19

Hinweis: Ihr Algorithmus kann bei der Suche nach einem Weg in einen Zyklus des Labyrinths laufen.

`plot_maze(maze, path)` führt das Labyrinth mit dem Array `path` zusammen und zeigt den zurückgelegten Weg rot an.
Speichern Sie die Bilder mithilfe der Funktion `plt.imsave('img.png',img)` ab. 

Hier wird das Labyrinth importiert und angezeigt:

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import sys

maze = np.array(Image.open('maze.png'))

plt.imshow(maze, cmap='gray')


# Unter Umständen muss die maximale Rekursionstiefe erweitert werden
print(sys.getrecursionlimit())
sys.setrecursionlimit(5000)
```

```python
def plot_maze(maze, path):
    rgb_maze = np.repeat(maze[:,:,None], 3, axis=2)
    rgb_maze[path==True,1] = 0
    rgb_maze[path==True,2] = 0
    return rgb_maze
path = np.zeros(maze.shape)
plt.imshow(plot_maze(maze, path))
```
