## 1. "Hello World"

In der Informatik ist das "Hello World" Programm traditionellerweise das allererste Programm, das man in einer neuen Programmiersprache schreibt. Die einzige Aufgabe des "Hello World" Programm ist es, den Text "Hello World" am Bildschirm auszugeben.

Das Ziel dieses Programms ist es, auf möglichst einfache Weise zu zeigen, welche Bestandteile notwendig sind, um ein vollständiges Programm zu schreiben. Dieses Programm verwendet man weiters auch dazu, um die Toolchain der jeweiligen Programmiersprache kennenzulernen.

Das "Hello World" Programm in Python schaut so aus:

````python
print("Hello World")
````

Wie man sehen kann, ist es in Python für ein vollständiges Programm ausreichend, die Befehle einfach hinzuschreiben. Der Python-Interpreter arbeitet dann die angegeben Befehle der Reihe nach ab.

In Java sähe das ganze wesentlich komplizierter aus:

```java
public class HelloWorld { 
      public static void main (String[] args) {
            System.out.println("Hello World!");
      }
}
```

## 2. Variablen

Variablen müssen in Python nicht explizit deklariert und definiert werden. Bei der ersten Zuweisung (der sogenannten *Initialisierung*) werden nicht existierende Variablen automatisch angelegt.

Python ist eine sogenannte **dynamisch typisierte** Sprache, d.h. die Datentypen von Variablen müssen nicht explizit angegeben werden, sondern werden automatisch anhand der in einer Variablen gespeicherten Inhalte hergeleitet. Eine Variable kann auch den Datentyp jederzeit wechseln.

Für Variablennamen gibt es in Python folgende Regeln:
- Variablennamen können Buchstaben (a-z, A-Z), Zahlen (0-9) oder Unterstriche (_) enthalten.
- Variablennamen müssen entweder mit einem Buchstaben (a-z, A-Z) oder mit einem Unterstrich (_) beginnen.
- Variablennamen können beliebig lang sein.
- Reservierte Schlüsselwörter wie ```if```, ```for```, ```def``` usw. dürfen nicht als Variablennamen verwendet werden ([hier](https://www.programiz.com/python-programming/keyword-list) eine Liste aller reservierten Schlüsselwörter).

Um Variablen Werte zuzuweisen, verwendet man den Zuweisungsoperator ```=```.

Beispiele, wie man Variablen verwendet:

````python
# Hier weisen wir der Variable a den Wert 34 zu.
# Die Variable a existiert noch nicht, und wird daher automatisch angelegt.
a = 34 
````

````python
# Hier sehen wir, dass die Variable a den Wert 34 angenommen hat.
print(a)
````

````python
# Der Datentyp der Variable a wird durch den Inhalt bestimmt
# in diesem Beispiel ist a vom Typ "int", was Ganzzahl bedeutet.
print(type(a))
````

````python
# Hier weisen wir a eine Gleitkommazahl zu.
a = 33.4
````

````python
# Hier sehen wir, dass die Variable a den Wert 33.4 angenommen hat.
print(a)
````

````python
# Der Datentyp von a passt sich dem Inhalt an, 
# und ist hier "float", was für Gleitkommazahl steht.
print(type(a))
````

````python
# Wir können auch Zeichenketten in a speichern.
a = "Hello World"
````

````python
# Hier sehen wir, dass die Variable a den Wert "Hello World" angenommen hat.
print(a)
````

````python
# Und wieder passt sich der Datentyp von a entsprechend an ("str" steht für string bzw. Zeichenkette).
print(type(a))
````

````python
# Weisen wir a wieder eine Zahl zu.
a = 5

# Und nun machen wir eine Rechnung mit dem Inhalt von a.
print(a + 6)
````

Hier sehen wir, was passiert, wenn wir nicht initialisierte Variablen verwenden:

````python
# Wenn wir Variablen verwenden, bevor wir ihnen einen Wert zugewiesen haben, gibt es eine Fehlermeldung.
# Die Fehlermeldung sagt uns, in welcher Zeile der Fehler aufgetreten ist und durch was er verursacht wurde.
print(a + c)

#  Wenn ein Fehler auftritt, wird die Ausführung unterbrochen, 
# d.h. der print Befehl hier wird solange nicht ausgeführt, bis wir den obrigen Fehler behoben haben.
print("Hallo") 
````

Und hier sehen wir, dass Variablen mit gleichen Namen, aber anderen Scope sich gegenseitig nicht beeinflussen:

````python
# Hier weisen wir a den Wert 3 zu
a = 3
# Und geben ihn aus
print("before a5(): a =", a) 

# Hier definieren wir eine Funktion a5(), die a den Wert 5 zuordnet und diesen Wert ausgibt
def a5():
    a = 5
    print("a5(): a =", a)

# Hier rufen wir die Funktion a5() auf
a5()

# Und hier sehen wir, dass a noch immer den Wert 3 hat.
# Die Variable a in der Funktion a5() hat einen anderen Scope, 
# und beeinflusst daher diese Variable a nicht
print("after a5(): a =", a) 
````

Und hier sehen wir, was passiert, wenn wir Variablen mit nicht kompatiblen Datentypen miteinander verwenden:

````python
a = 5 # Variable a enthält eine Zahl
b = "Hallo" # Variable b enthält eine Zeichenkette

# a und b enthalten inkompatible Datentypen, und wir erhalten deshalb eine Fehlermeldung
a + b 
````

````python
# Das Ergebnis dieser Division ist 2.25, eine Gleitkommazahl
9 / 4
````

````python
# Bei der obrigen ganzzahligen Division fehlt noch der sogenannte Rest, 
# den erhalten wir durch den Modulo Operator %
9 % 4
````

### 3.2. Logische Verknüpfungen

In Python steht **True** für den Wahrheitswert wahr, und **False** für den Wahrheitswert falsch. Python unterscheidet zwischen Groß- und Kleinschreibung (das nennt man "case-sensitive"), daher immer darauf achten das der erste Buchstabe von True und False groß geschrieben ist.

Python kennt folgende logische Verknüpfungen:
- ```and```
- ```or```
- ```not```

Ausserdem können Klammern in logischen Ausdrücken verwendet werden.

Alle anderen logischen Operatoren (wie z.B. ```xor```, ```^```) müssen durch Ersatzausdrücke, die nur aus den oben genannten Operatoren bestehen, ersetzt werden.

Beispiele für logische Ausdrücke:

````python
# Hier definieren wir uns ein paar Variablen, die Wahrheitswerte enthalten.
# Wir könnten auch logische Ausdrücke direkt mit True/False bilden, in der Praxis kommen solche Ausdrücke aber kaum vor.
a = True
b = False
c = True
````

````python
# Und-Verknüpfung
print(a and b)
````

````python
# Oder-Verknüpfung
print(a or b)
````

````python
# Verneinung
print(not a)
````

````python
# Etwas komplizierterer Ausdruck, der Klammern verwendet
print((a or b) and c)
````

````python
# Python kennt kein xor, daher müssen wir einen Ersatzausdruck verwenden
print(a and not b or not a and b)

print(a != b) #funktioniert nur, falls a und b tatsächlich Boolsche Werte sind (bzw. alle jeweiligen True- und False- Werte identisch sind)
````

Es ist auch möglich andere Werte miteinander zu vergleichen: 
- Bei Zahlen ist 0 False während alle anderen Zahlen den Wert True annehmen
- Bei Strings ist ein leerer String ```""``` False, während jedes Zeichen dem String den Wert True zuordnet.

````python
a = 0
b = 1
print(a and b)
print(a or b)
````

#### 3.2.1. Bitweise Operatoren

Die logischen Verknüpfungen werden auch auf Bit-Ebene verwendet. Hierbei ist es hilfreich die Zahlen im Binärsystem oder Hexadezimalsystem zu betrachten. Dafür wird ein Präfix vorangestellt.

Die Operatoren vergleichen einzelne Bits zweier Zahlen miteinander:

Operator $\;$ Beschreibung \
```&``` $\;\;\;\;\;\;\;\;\:\:$ Bitwise and \
```|``` $\;\;\;\;\;\;\;\;\;\:$ Bitwise OR \
```^``` $\;\;\;\;\;\;\;\;\;\:$ Bitwise XOR (exclusive OR) \
```~``` $\;\;\;\;\;\;\;\;\:\:$ Bitwise NOT \
```<<``` $\;\;\;\;\;\;\;\;$ Bitwise left shift \
```>>``` $\;\;\;\;\;\;\;\;$ Bitwise right shift


````python
# Binärzahlen haben den Präfix 0b, Hexadezimalzahlen den Präfix 0x
print(0b1000)
print(0xFF)
````

````python
# Bitwise and
print(bin(0b1100 & 0b1010))
````

````python
# Bitwise or
print(bin(0b1100 | 0b1010))
````

````python
# Bitwise xor
print(bin(0b0110 ^ 0b1010))
````

Bei bitweisen Operationen mit Negativzahlen ist ein wenig Vorsicht geboten, da diese im Zweierkomplement gespeichert sind.
Zur Umrechnung einer negativen Zahl wird diese invertiert und 1 dazugezählt.

````python
# Bitwise not
print(bin(~0b0110))

# Da das erste Bit das Vorzeichen ist, kann man das Ergebnis mit einer &-Operation wieder ins positive konvertieren
# 0xF ist ist hierbei von gleicher Länge wie der Input
bin((~0b0110)&0xF)
````

````python
print(bin(0b1100 & 0b1010))
print(bin(0b1100 & -0b1010)) #1100 & 0110
````

````python
# Bitwise left shift 
print(bin(0b0001<<2))
````

````python
# Bitwise right shift 
print(bin(0b101>>2))
````

### 3.3. Vergleichsoperatoren

Um Werte vergleichen zu können, gibt es in Python auch sogenannte Vergleichsoperatoren:\
```==``` $\;\;\;\;\;\;\;\;\:\:$ gleich\
```!=``` $\;\;\;\;\;\;\;\;\:\:$ ungleich\
```>``` $\;\;\;\;\;\;\;\;\;\;\;$ größer\
```>=``` $\;\;\;\;\;\;\;\;\:\:$ größer oder gleich\
```<``` $\;\;\;\;\;\;\;\;\;\;\;$ kleiner\
```<=``` $\;\;\;\;\;\;\;\;\:\:$ kleiner oder gleich

Hier sehen wir Beispiele, die diese Vergleichsoperatoren verwenden:

````python
# 3 und 2 sind nicht gleich, daher bekommen wir False als Ergebnis
# Bitte unbedingt beachten, dass der Gleichheitsoperator aus ZWEI '=' besteht
print(3 == 3)
````

````python
# 3 und 2 sind nicht gleich, daher bekommen wir hier True als Ergebnis
print(3 != 2)
````

````python
# 3 ist größer als 2, daher bekommen wir hier True als Ergebnis
print(3 > 2)
````

````python
# 3 ist gleich 3, daher bekommen wir hier True als Ergebnis
print(3 >= 3)
````

````python
# wenn wir nur den größer-Operator verwenden, dann bekommen wir False als Ergebnis
print(3 > 3)
````

````python
# 3 ist nicht kleiner als 2, daher bekommen wir hier False als Ergebnis
print(3 < 2)
````

### 3.4. Funktionsaufrufe

Funktionsaufrufe in Python schauen folgendermaßen aus:

`<Funktionsname>(<Parameterliste>)`

Der Funktionsname identifiziert die Funktion, die wir aufrufen wollen. Mit den Klammern ```()``` teilen wir Python mit, dass dieser Befehl ein Funktionsaufruf ist, und eventuelle Eingabeparameter stehen innerhalb der Klammern.

Funktionen können einen Rückgabewert haben, diesen Wert können wir entweder mit dem Zuweisungsoperator in einer Variablen speichern, oder direkt innerhalb eines Ausdruckes verwenden. In diesem Fall schreiben wir den Funktionsaufruf direkt an die Stelle in unserem Ausdruck, wo wir den Rückgabewert verwenden wollen.

Wir können den Rückgabewert auch einfach ignorieren, dann wird er einfach verworfen.

````python
# Hier definieren wir uns eine Funktion double, die einen Eingabeparameter a hat, 
# und der Rückgabewert dieser Funktion ist 2*a.
def double(a):
    return 2*a
````

````python
# Hier rufen wir unsere Funktion mit dem Eingabeparameter 4 auf, und erhalten 8 als Ergebis
print(double(16.4568))
````

````python
# Hier speichern wir das Ergebnis des Funktionsaufrufs in der Variable a
a = double(4)
# Welche nun den Wert 8 enthält
print(a)
````

````python
# Wir können Funktionsaufrufe auch direk in arithmetischen Ausdrücken verwenden.
# In diesem Fall wird mit dem Rückgabewert gerechnet.
print(double(1) + double(4))
````

````python
# Wir können auch Funktionsaufrufe als Parameter eines anderen Funktionsaufrufes verwenden.
# In diesem Fall dient der Rückgabewert des einen Funktionsaufrufs als Eingabeparameter des anderen Funktionsaufruf.
print(double(double(4)))
````

### 3.4.1. Call by value/reference

Auch in Python gibt es die Unterscheidung zwischen *call-by-value* und *call-by-reference*. 

Wie die meisten Skriptsprachen verwendet Python für simple Datentypen *call-by-value*, und für komplexe Datentypen *call-by-reference*.

Hier sieht man ein Beispiel für *call-by-value*:

````python
# Die Funktion callbyvalue() überschreibt den Eingabeparameter a mit einem anderen Wert
def callbyvalue(a):
    print("callbyvalue() before a=66: a =", a)
    a = 66
    print("callbyvalue() after a=66: a =", a)

a = 34
print("a =", a)

callbyvalue(a)

print("a =", a)
````

`a` ist also vor und nach der Ausführung von `callbyvalue()` identisch, das `a` innerhalb der Funktion ist unabhängig vom äußeren `a`.

Und hier sieht man ein Beispiel für *call-by-reference*:

````python
# Die Funktion callbyreference() hängt an eine übergebene Liste ein neues Element dran
def callbyreference():
    l.append(32)
    print("callbyreference(): l =", l)


l =  []
print("l =", l)


callbyreference()


print("l =", l)
````

`l` ist hier eine Liste, eine komplexere Datenstruktur mit der sich mehrere Werte zusammenfassen lassen. Diese ist nicht vor und nach der Ausfühung durch `callbyreference()` identisch: die Endung `.append` hat `l` ein Listenelement mit dem Wert 32 angehängt. Bei komplexen Datentypen beeinflusst also das Innenleben der Funktion das äußere.

### 3.5. Sprachenspezifische Befehle

Im Moment ist für uns nur vor Allem ein Python-spezifischer Befehl relevant, und das ist die `print` Funktion.

`print` kann beliebig viele Eingabeparameter haben, und gibt diese Eingabeparameter bzw. deren Werte am Bildschirm aus.

````python
# print mit nur einem Eingabeparameter
print("Hallo Welt")
````

````python
# Dieselbe Ausgabe wie oben mit zwei Eingabeparametern (das Leerzeichen wird automatisch hinzugefügt)
print("Hallo", "Welt")
````

````python
# print kann auch Zahlen ausgeben, hier wird die Zahl 42 ausgegeben
print(42)
````

````python
# Unterschiedliche Typen kombiniert
print("Die Antwort auf das Leben, das Universum und den ganzen Rest ist", 42)
````

````python
# Man kann auch den Wert von Variablen ausgeben
a = 42.42
print("a =", a)
````

````python
# Hier haben wir einen Funktionsaufruf von print mit 5 Eingabeparametern
print("Hallo", 32, 43.2, "bla", True) 
````

Des Weiteren wird noch die ```range```-Funktion relevant. Diese gibt eine ganzzahlige Zahlensequenz aus. 
Für Kommazahlen siehe ```np.arange```.

```range``` hat 3 Eingabeparameter, von denen allerdings nur einer benötigt wird.
- ```start```: Anfangswert der Sqeuenz (optional)
- ```stop```: erster Wert nach Endwert der Sequenz
- ```step```: Schrittweite

````python
# Mit nur dem Eingabeparameter stop werden Werte von 0 bis stop-1 ausgegeben
sequence = range(5)
print(sequence)

# Mit dem Asterisk als Präfix lassen sich die Werte der Sequenz anzeigen.
print(*sequence)
````

````python
# Mit einem zweiten Argument lässt sich der Startpunkt definieren.
print(*range(3, 8))
````

````python
# Auch negative Zahlen sind möglich
print(*range(-5, 5))
````

````python
# Ein drittes Argument definiert die Schrittweite. Auch hier ist der Endwert nicht eingeschlossen.
print(*range(0, 12, 3))
````

````python
# Es ist auch möglich rückwärts zu zählen.
print(*range(5, 0, -1))
````
